```python
dataset_path = "/content/coin_dataset/Indian Coin Currency Dataset"

img_size = (128, 128)
batch_size = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=img_size,
    batch_size=batch_size
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=img_size,
    batch_size=batch_size
)

print("Classes:", train_ds.class_names)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

model = models.Sequential([

    # Input image
    layers.Input(shape=(128, 128, 3)),

    # First CNN layer
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    # Second CNN layer
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    # Third CNN layer
    layers.Conv2D(128, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    # Convert feature maps into one long vector
    layers.Flatten(),

    # Fully connected layer
    layers.Dense(128, activation="relu"),

    # 4 coin classes
    layers.Dense(4, activation="softmax")
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully!")
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10
)
```

```python
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1)
])
```

```python
model = models.Sequential([

    # Input image
    layers.Input(shape=(128, 128, 3)),

    # Data augmentation
    data_augmentation,

    # First CNN layer
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    # Second CNN layer
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    # Third CNN layer
    layers.Conv2D(128, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    # Convert feature maps to one vector
    layers.Flatten(),

    # Fully connected layer
    layers.Dense(128, activation="relu"),

    # Reduce overfitting
    layers.Dropout(0.5),

    # 4 coin classes
    layers.Dense(4, activation="softmax")
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(128, 128, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze the pretrained model
base_model.trainable = False

model = models.Sequential([
    layers.Input(shape=(128, 128, 3)),

    # Convert pixel values from 0-255 to -1 to 1
    layers.Rescaling(1./127.5, offset=-1),

    # Small image variations
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),

    # Pretrained MobileNetV2
    base_model,

    layers.GlobalAveragePooling2D(),

    layers.Dropout(0.3),

    # 4 coin classes
    layers.Dense(4, activation="softmax")
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10
)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.utils import load_img, img_to_array

# Add your image path here
image_path = "image1.jfif"

# Load image
img = load_img(image_path, target_size=(128, 128))

# Display image
plt.figure(figsize=(5, 5))
plt.imshow(img)
plt.title("New Coin Image")
plt.axis("off")
plt.show()

# Prepare image
img_array = img_to_array(img)
img_array = np.expand_dims(img_array, axis=0)

# Predict
predictions = model.predict(img_array)[0]

class_names = train_ds.class_names

predicted_index = np.argmax(predictions)

print("Predicted Coin:", class_names[predicted_index])
print("Confidence:", f"{predictions[predicted_index] * 100:.2f}%")

# All probabilities
for coin, probability in zip(class_names, predictions):
    print(f"{coin}: {probability * 100:.2f}%")
```

```python
probabilities = predictions * 100

plt.figure(figsize=(8, 5))

bars = plt.bar(class_names, probabilities)

plt.title("Coin Prediction Probability")
plt.xlabel("Coin")
plt.ylabel("Probability (%)")

plt.ylim(0, max(100, max(probabilities) + 10))

for bar, value in zip(bars, probabilities):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 1,
        f"{value:.2f}%",
        ha="center"
    )

plt.show()
```
